# Practical 13: Text Data Cleaning & Custom Tokenization using Python

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to clean raw, noisy text data: convert characters to lowercase using lower(), remove special punctuation symbols using replace(), eliminate redundant whitespace using strip(), and tokenize sentences into individual word tokens using split().

```python
import string

raw_texts = [
    "   Welcome to the BIG-DATA lab!!!   We love   Spark, Pandas & PySpark...  ",
    "Machine learning (ML) models are TRAINED on huge datasets; accuracy matters!!  ",
    "   Contact: student@college.edu - Batch #DS505 @Surat   ",
]
for t in raw_texts:
    print(repr(t))

def clean_text(text: str) -> str:
    text = text.lower()
    for p in string.punctuation:
        text = text.replace(p, "")
    text = " ".join(text.split())
    return text.strip()

def tokenize(text: str) -> list:
    return text.split()

for i, raw in enumerate(raw_texts, 1):
    cleaned = clean_text(raw)
    tokens = tokenize(cleaned)
    print(f"--- Sentence {i} ---")
    print("Raw     :", repr(raw))
    print("Cleaned :", repr(cleaned))
    print("Tokens  :", tokens)
    print("Count   :", len(tokens), "\n")
```
