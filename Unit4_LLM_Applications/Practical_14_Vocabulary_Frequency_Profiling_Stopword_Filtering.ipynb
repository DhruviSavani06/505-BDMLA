# Practical 14: Vocabulary Frequency Profiling & Stopword Filtering

**UNIT 4: Introduction to Large Language Models and Big Data Applications**

**Problem Statement:** Write a Python program to profile vocabulary in a text corpus: clean sentences, filter out common English stopwords (e.g., 'the', 'is', 'in', 'and'), calculate word occurrence frequencies using collections.Counter, and display the top 5 most frequent keywords.

```python
import string
from collections import Counter

corpus = [
    "Cloud computing gives teams elastic storage for large datasets.",
    "The analytics team cleans the data and builds a dashboard for the managers.",
    "Spark jobs run faster when the data is partitioned across many nodes.",
    "Good features help a model learn useful patterns from the data.",
    "Students use Python and Pandas to explore big data in the lab.",
]

STOPWORDS = {"the", "is", "in", "and", "of", "to", "for", "it", "on", "are", "a", "an", "over", "use", "with", "that", "this", "as", "at", "by", "from", "can"}

def clean(sentence: str) -> str:
    sentence = sentence.lower()
    for p in string.punctuation:
        sentence = sentence.replace(p, "")
    return " ".join(sentence.split())

all_tokens = []
for s in corpus:
    all_tokens.extend(clean(s).split())

filtered = [w for w in all_tokens if w not in STOPWORDS]

print(f"Total tokens (before filtering): {len(all_tokens)}")
print(f"Total tokens (after  filtering): {len(filtered)}")
print(f"Vocabulary size (unique words) : {len(set(filtered))}")

freq = Counter(filtered)
print(freq)

for rank, (word, count) in enumerate(freq.most_common(5), 1):
    print(f"{rank}. {word:<10} {count}  {'#' * count}")
```
