# Practical 01: Dataset Ingestion, Inspection & Descriptive Statistics with Pandas

**UNIT 1: Introduction to Big Data and Large Dataset Handling**

**Problem Statement:** Write a Python program using Pandas to load a structured dataset (CSV/JSON), display its first 5 rows, last 5 rows, dimensions (shape), column names, and data types with info(). Generate comprehensive summary statistics of all numeric and categorical attributes using describe(include='all').

```python
import io
import numpy as np
import pandas as pd

pd.set_option("display.max_columns", None, "display.width", 200)

gen = np.random.default_rng(7)
n = 150
raw = pd.DataFrame({
    "Order ID": np.arange(1001, 1001 + n),
    "Customer Name": [f"Client_{i:03d}" for i in gen.integers(1, 45, n)],
    "City": gen.choice(["Delhi", "Jaipur", "Surat", "Indore", "Nagpur", "Pune"], n),
    "Product Category": gen.choice(["Electronics", "Clothing", "Groceries", "Toys", "Books"], n),
    "Quantity": gen.integers(1, 12, n),
    "Unit Price": np.round(gen.uniform(150, 4500, n), 2),
    "Discount %": gen.choice([0, 5, 10, 15, 25], n),
    "Order Date": pd.date_range("2025-03-01", periods=n, freq="D").strftime("%Y-%m-%d"),
})

csv_file = io.StringIO()
raw.to_csv(csv_file, index=False)
csv_file.seek(0)
json_file = io.StringIO(raw.to_json(orient="records"))

orders = pd.read_csv(csv_file)
print("Dataset loaded successfully!")

print(orders.head())

print(orders.tail())

print("Shape (rows, columns):", orders.shape)
print("Column names:", orders.columns.tolist())

orders.info()

print(orders.describe(include='all'))

df_json = pd.read_json(json_file)
print("JSON load – shape:", df_json.shape)
print(df_json.head())
```
