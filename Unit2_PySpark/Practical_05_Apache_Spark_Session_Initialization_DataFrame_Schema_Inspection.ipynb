# Practical 05: Apache Spark Session Initialization & DataFrame Schema Inspection

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a Python program to install and initialize an Apache Spark session in PySpark using SparkSession.builder. Create a structured PySpark DataFrame from Python dictionaries, display its schema tree hierarchy using printSchema(), and print its rows using .show().

```python
from pyspark.sql import SparkSession, Row

spark = (SparkSession.builder
         .appName("DS505-Practical05")
         .master("local[*]")
         .config("spark.ui.showConsoleProgress", "false")
         .getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

print("Spark version:", spark.version)

data = [
    {"Name": "Ananya", "Department": "IT", "City": "Vadodara", "Salary": 69000, "Experience": 4},
    {"Name": "Yash", "Department": "HR", "City": "Surat", "Salary": 51000, "Experience": 2},
    {"Name": "Pooja", "Department": "Finance", "City": "Rajkot", "Salary": 63000, "Experience": 7},
    {"Name": "Harsh", "Department": "IT", "City": "Surat", "Salary": 85000, "Experience": 8},
    {"Name": "Simran", "Department": "Sales", "City": "Pune", "Salary": 46000, "Experience": 1},
]
df = spark.createDataFrame([Row(**d) for d in data])
df.printSchema()
df.show()
spark.stop()
```
