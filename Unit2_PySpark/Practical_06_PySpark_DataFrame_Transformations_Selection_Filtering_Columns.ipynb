# Practical 06: PySpark DataFrame Transformations: Selection, Filtering & Columns

**UNIT 2: Big Data Processing using PySpark**

**Problem Statement:** Write a PySpark program to execute DataFrame transformations: select specific columns using .select(), filter rows using multi-condition boolean logic (e.g., Salary > 60000 and City == 'Surat') using .filter(), create a computed column using .withColumn(), and rename a column using .withColumnRenamed().

```python
from pyspark.sql import SparkSession, Row
from pyspark.sql.functions import col

spark = (SparkSession.builder.appName("DS505-Practical06").master("local[*]")
         .config("spark.ui.showConsoleProgress", "false").getOrCreate())
spark.sparkContext.setLogLevel("ERROR")

data = [
    {"Name": "Ananya", "Department": "IT", "City": "Vadodara", "Salary": 69000},
    {"Name": "Yash", "Department": "HR", "City": "Surat", "Salary": 51000},
    {"Name": "Pooja", "Department": "Finance", "City": "Rajkot", "Salary": 63000},
    {"Name": "Harsh", "Department": "IT", "City": "Surat", "Salary": 85000},
    {"Name": "Simran", "Department": "Sales", "City": "Pune", "Salary": 46000},
    {"Name": "Dev", "Department": "IT", "City": "Surat", "Salary": 93000},
]
df = spark.createDataFrame([Row(**d) for d in data])
df.show()

df.select("Name", "City", "Salary").show()

filtered = df.filter((col("Salary") > 60000) & (col("City") == "Surat"))
filtered.show()

with_bonus = df.withColumn("Annual_Bonus", col("Salary") * 0.10)
with_bonus.show()

renamed = with_bonus.withColumnRenamed("Name", "Employee_Name")
renamed.show()
renamed.printSchema()
spark.stop()
```
